แปลงระยะทางมามหาวิทยาลัย เป็นกิโลเมตรและเก็บไว้ใน 'ระยะทาง(กม.)'

```python
import re

for i in range(len(df)):
    text = str(df.loc[i, 'ระยะทางมามหาวิทยาลัย']).strip().lower()

    text = text.replace(' ', '').replace('กม.', 'กม').replace('เมตร', 'm')

    num_list = re.findall(r'[\d.]+', text)
    if not num_list:
        continue

    num = float(num_list[0])

    if 'km' in text or 'กม' in text or 'กิโล' in text:
        df.loc[i, 'ระยะทาง(กม.)'] = round(num, 2)
    elif 'm' in text:
        df.loc[i, 'ระยะทาง(กม.)'] = round(num / 1000, 2)
```

```python
positive_words = [
    "ดี", "ดีมาก", "น่ารัก", "ประทับใจ", "สะอาด", "กว้าง",
    "ปลอดภัย", "สบาย", "น่าอยู่", "คุ้มค่า", "ราคาดี", "ย่อมเยา",
    "ดูแลดี", "บริการดี", "ใจดี", "เฟรนด์ลี่"
]

negative_words = [
    "แย่", "ไม่ดี", "เสียงดัง", "เหม็นบุหรี่", "เก่า", "แพง",
    "คืนยาก", "ไม่คืน", "ไม่ค่อยเก็บเสียง", "ค้างบ่อย", "ไม่สะดวก"
]
```

```python
def get_sentiment(text):
    t = str(text)
    score = 0
    for w in positive_words:
        if w in t:
            score += 1
    for w in negative_words:
        if w in t:
            score -= 1
    if score > 0:
        return "positive"
    elif score < 0:
        return "negative"
    else:
        return "neutral"

def get_aspect(text):
    t = str(text)
    if any(w in t for w in ["สะอาด", "ฝุ่น", "แม่บ้าน"]):
        return "ความสะอาด"
    if any(w in t for w in ["ปลอดภัย", "ยาม", "รปภ", "กล้องวงจรปิด"]):
        return "ความปลอดภัย"
    if any(w in t for w in ["ใกล้มอ", "ใกล้มหาวิทยาลัย", "เดินไป", "ซอย", "ของกิน", "ร้านอาหาร", "เซเว่น"]):
        return "ทำเล/ของกิน"
    if any(w in t for w in ["ราคา", "ถูก", "แพง", "ค่าหอ", "มัดจำ", "ค่ามัดจำ"]):
        return "ราคา"
    if any(w in t for w in ["เสียงดัง", "เงียบ", "ไม่ค่อยเก็บเสียง"]):
        return "เสียง/ความสงบ"
    if any(w in t for w in ["ที่จอดรถ", "ลิฟต์", "ตู้เย็น", "แอร์", "ส่วนกลาง", "เฟอร์นิเจอร์"]):
        return "สิ่งอำนวยความสะดวก"
    return "อื่นๆ"

def make_summary(text):
    t = str(text).strip()
    if len(t) <= 60:
        return t
    return t[:60] + "..."

df["summary"] = df["รีวิว"].apply(make_summary)
df["sentiment"] = df["รีวิว"].apply(get_sentiment)
df["aspect"] = df["รีวิว"].apply(get_aspect)

df.head()
```

```python
def get_aspect_en(text):
    t = str(text)

    cleanliness_words = ["สะอาด", "แม่บ้าน", "ห้องรก", "ห้องหรู"]
    safety_words = ["ปลอดภัย", "ยาม", "รปภ", "กล้องวงจรปิด"]
    location_words = ["ใกล้", "ใกล้มอ", "หน้ามอ", "เดินไป", "เดินข้ามถนน", "ซอย", "กลางซอย", "ระยะทาง"]
    price_words = ["ถูก", "แพง", "คุ้มค่า", "ราคาดี", "ราคาย่อมเยา", "ค่าไฟ", "ค่าหอ", "มัดจำ", "ค่ามัดจำ"]
    noise_words = ["เสียงดัง", "ไม่ค่อยเก็บเสียง", "เงียบ", "สงบ", "วุ่นวาย"]
    facility_words = ["ที่จอดรถ", "ลิฟต์", "ลิฟท์", "ตู้เย็น", "แอร์", "ส่วนกลาง", "เฟอร์นิเจอร์", "โต๊ะ", "อินเทอร์เน็ต", "Wi-Fi", "wifi"]
    service_words = ["เจ้าของหอ", "เจ้าของ", "แม่บ้าน", "คนดูแล", "พนักงาน", "นิติ", "บริการดี", "ดูแลดี", "ใจดี"]

    if any(w in t for w in cleanliness_words):
        return "cleanliness"
    if any(w in t for w in safety_words):
        return "safety"
    if any(w in t for w in location_words):
        return "location_access"
    if any(w in t for w in price_words):
        return "price_value"
    if any(w in t for w in noise_words):
        return "noise"
    if any(w in t for w in facility_words):
        return "facilities"
    if any(w in t for w in service_words):
        return "staff_service"

    return "other"
df["aspect_en"] = df["รีวิว"].apply(get_aspect_en)
```
